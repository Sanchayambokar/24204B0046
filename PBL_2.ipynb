1. List Operations
Develop a Python program to create, access, update, and delete student records using a List data structure.

```python
students = []

while True:
    print("\n--- Menu ---")
    print("1. Add Student")
    print("2. Display Students")
    print("3. Update Student")
    print("4. Delete Student")
    print("5. Exit")

    choice = int(input("Enter choice: "))
    if choice == 1:
        roll = int(input("Enter Roll No: "))
        name = input("Enter Name: ")
        students.append([roll, name])
        print("Student added")
    elif choice == 2:
        print("\nStudent Records:")
        for s in students:
            print("Roll:", s[0], " Name:", s[1])
    elif choice == 3:
        roll = int(input("Enter Roll No to update: "))
        for s in students:
            if s[0] == roll:
                s[1] = input("Enter new name: ")
                print("Record updated")
                break
    elif choice == 4:
        roll = int(input("Enter Roll No to delete: "))
        for s in students:
            if s[0] == roll:
                students.remove(s)
                print("Record deleted")
                break
    elif choice == 5:
        print("Program ended")
        break
    else:
        print("Invalid choice")
```

2. Tuple Operations
Write a Python program to store fixed laboratory information using a Tuple and perform access and modification operations.

```python
lab_info = ("Electronics Lab", "Room 101", 30, "Ms. Patil")

print("Original Tuple:")
print(lab_info)

print("\nAccessing elements:")
print("Lab Name:", lab_info[0])
print("Room Number:", lab_info[1])
print("Capacity:", lab_info[2])
print("Instructor:", lab_info[3])

lab_list = list(lab_info)
lab_list[3] = "Mr. Sharma"
lab_info = tuple(lab_list)

print("\nModified Tuple:")
print(lab_info)
```

4. Dictionary Operations
Write a Python program to store and manage student details (Roll No, Name, Marks) using a Dictionary, including create, access, update, delete, and traversal operations.


```python
students = {}

students[1] = {"Name": "Rahul", "Marks": 85}
students[2] = {"Name": "Aman", "Marks": 90}

print("All Records:")
print(students)

print("\nAccess Record (Roll No 1):")
print(students[1])

students[1]["Marks"] = 95
print("\nAfter Update:")
print(students)

del students[2]
print("\nAfter Delete:")
print(students)

print("\nTraversal:")
for roll, details in students.items():
    print("Roll No:", roll, "Name:", details["Name"], "Marks:", details["Marks"])
```

5. Student Record Management
Design a Python program using appropriate data structures to maintain and display multiple student records.

```python
students = []

while True:
    print("\n1. Add Student")
    print("2. Display Students")
    print("3. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        roll = int(input("Enter Roll No: "))
        name = input("Enter Name: ")
        marks = float(input("Enter Marks: "))
        students.append({"Roll": roll, "Name": name, "Marks": marks})

    elif choice == 2:
        for s in students:
            print("Roll:", s["Roll"], "Name:", s["Name"], "Marks:", s["Marks"])

    elif choice == 3:
        break

    else:
        print("Invalid choice")
```

6. Laboratory Record Handling
Develop a Python program to manage laboratory resources and availability using Python data structures.

```python
lab_resources = {}

while True:
    print("\n1. Add Resource")
    print("2. Display Resources")
    print("3. Update Availability")
    print("4. Delete Resource")
    print("5. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        name = input("Enter resource name: ")
        status = input("Enter availability (Available/Not Available): ")
        lab_resources[name] = status

    elif choice == 2:
        for r, s in lab_resources.items():
            print("Resource:", r, "Status:", s)

    elif choice == 3:
        name = input("Enter resource name to update: ")
        if name in lab_resources:
            lab_resources[name] = input("Enter new availability: ")

    elif choice == 4:
        name = input("Enter resource name to delete: ")
        if name in lab_resources:
            del lab_resources[name]

    elif choice == 5:
        break

    else:
        print("Invalid choice")
```

7. Data Traversal
Write a Python program to traverse and display all stored student and laboratory records efficiently.



```python
students = [
    {"Roll": 1, "Name": "Rahul", "Marks": 85},
    {"Roll": 2, "Name": "Sanchay", "Marks": 90}
]

lab_resources = {
    "Oscilloscope": "Available",
    "Multimeter": "Not Available",
    "Power Supply": "Available"
}

print("Student Records:")
for s in students:
    print("Roll:", s["Roll"], "Name:", s["Name"], "Marks:", s["Marks"])

print("\nLaboratory Resources:")
for r, status in lab_resources.items():
    print("Resource:", r, "Status:", status)
```
