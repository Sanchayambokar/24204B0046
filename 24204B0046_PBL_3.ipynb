2. User-Defined Functions

Develop a Python program using user-defined functions to implement Half Adder and Full Adder logic.


```python
def half_adder(a, b):
    sum_bit = a ^ b
    carry = a & b
    return sum_bit, carry

def full_adder(a, b, cin):
    sum_bit = a ^ b ^ cin
    carry = (a & b) | (b & cin) | (a & cin)
    return sum_bit, carry

a = int(input("Enter first bit (0 or 1): "))
b = int(input("Enter second bit (0 or 1): "))

s, c = half_adder(a, b)
print("Half Adder -> Sum:", s, "Carry:", c)

cin = int(input("Enter carry in (0 or 1): "))

s, c = full_adder(a, b, cin)
print("Full Adder -> Sum:", s, "Carry:", c)
```

3. Classes and Objects

Write a Python program to define a Student class with appropriate data members and member functions.
Create objects of the class and display student details.


```python
class Student:
    def __init__(self, name, roll, marks):
        self.name = name
        self.roll = roll
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Roll No:", self.roll)
        print("Marks:", self.marks)
        print()

s1 = Student("Sanchay", 46, 97)
s2 = Student("Aman", 17, 90)

s1.display()
s2.display()
```


5. User-Defined Module

Create and use a user-defined module (for example, a calculator module) to perform arithmetic operations.

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    return a / b
```

```python
import calculator

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("Addition:", calculator.add(a, b))
print("Subtraction:", calculator.subtract(a, b))
print("Multiplication:", calculator.multiply(a, b))
print("Division:", calculator.divide(a, b))
```


6. NumPy Package

Write a Python program to demonstrate the use of the NumPy package for performing array operations.


```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Array 1:", a)
print("Array 2:", b)

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)

print("Mean of Array 1:", np.mean(a))
print("Max of Array 2:", np.max(b))
```

```python
import os
os.makedirs("mypackage", exist_ok=True)
```

```python
%%writefile mypackage/add.py

def add(a, b):
    return a + b
```
