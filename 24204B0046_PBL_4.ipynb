1. Linear Search
   
Develop a Python program to implementLinear Search to find a given element from a list of academic records.



```python
records = ["Aman", "Sanchay", "Swayam", "Rohit", "Rakshit"]

key = input("Enter the student name to search: ")

found = False

for name in records:
    if name == key:
        found = True
        break

if found:
    print("Record Found")
else:
    print("Record Not Found")
```

2. Binary Search
Develop a Python program to implementBinary Search on a sorted list of student roll numbers.

```python
roll_numbers = [1, 3, 5, 7, 9, 11, 13, 15]

key = int(input("Enter roll number to search: "))

low = 0
high = len(roll_numbers) - 1
found = False

while low <= high:
    mid = (low + high) // 2

    if roll_numbers[mid] == key:
        found = True
        break
    elif roll_numbers[mid] < key:
        low = mid + 1
    else:
        high = mid - 1

if found:
    print("Roll number found")
else:
    print("Roll number not found")
```

3. Bubble Sort
  
  Write a Python program to implement Bubble Sort to arrange marks of students in ascending or descending order.




```python
marks = list(map(int, input("Enter marks separated by space: ").split()))

order = input("Enter 'A' for Ascending or 'D' for Descending: ").upper()

n = len(marks)

for i in range(n):
    for j in range(0, n - i - 1):
        if (order == 'A' and marks[j] > marks[j + 1]) or (order == 'D' and marks[j] < marks[j + 1]):
            marks[j], marks[j + 1] = marks[j + 1], marks[j]

print("Sorted marks:", marks)
```

4. Selection Sort
   
Write a Python program to implementSelection Sort to sort laboratory equipment IDs.

```python
ids = list(map(int, input("Enter equipment IDs separated by space: ").split()))

n = len(ids)

for i in range(n):
    min_index = i
    for j in range(i + 1, n):
        if ids[j] < ids[min_index]:
            min_index = j
    ids[i], ids[min_index] = ids[min_index], ids[i]

print("Sorted Equipment IDs:", ids)
```

5. Insertion Sort

Write a Python program to implementInsertion Sort to organize student data efficiently.


```python
students = list(map(int, input("Enter student data separated by space: ").split()))

for i in range(1, len(students)):
    key = students[i]
    j = i - 1

    while j >= 0 and students[j] > key:
        students[j + 1] = students[j]
        j -= 1

    students[j + 1] = key

print("Sorted student data:", students)
```

6. Algorithm Comparison
   
Analyze and compare the performance ofsearching and sorting algorithms based on time complexity.


```python
import time

data = [45, 12, 78, 34, 23, 89, 67]
key = 34


def linear_search(arr, key):
    for i in arr:
        if i == key:
            return True
    return False


def binary_search(arr, key):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == key:
            return True
        elif arr[mid] < key:
            low = mid + 1
        else:
            high = mid - 1
    return False


def bubble_sort(arr):
    a = arr.copy()
    n = len(a)
    for i in range(n):
        for j in range(0, n-i-1):
            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]
    return a


start = time.time()
linear_search(data, key)
print("Linear Search Time:", time.time() - start)


sorted_data = bubble_sort(data)

start = time.time()
binary_search(sorted_data, key)
print("Binary Search Time:", time.time() - start)


start = time.time()
bubble_sort(data)
print("Bubble Sort Time:", time.time() - start)
```
