1. Stack Implementation

Write a Python program to implement a Stack using array with PUSH and POP operations for managing records.



```python
stack = []

def push():
    item = input("Enter record to push: ")
    stack.append(item)
    print("Pushed:", item)

def pop():
    if len(stack) == 0:
        print("Stack is empty")
    else:
        print("Popped:", stack.pop())

while True:
    print("\n1. PUSH")
    print("2. POP")
    print("3. DISPLAY")
    print("4. EXIT")

    choice = int(input("Enter choice: "))

    if choice == 1:
        push()
    elif choice == 2:
        pop()
    elif choice == 3:
        print("Stack:", stack)
    elif choice == 4:
        break
    else:
        print("Invalid choice")
```

2. Queue Implementation

Develop a Python program to implement a Queue using array with ENQUEUE and DEQUEUE operations for data processing.




```python
queue = []

def enqueue():
    item = input("Enter data to enqueue: ")
    queue.append(item)
    print("Enqueued:", item)

def dequeue():
    if len(queue) == 0:
        print("Queue is empty")
    else:
        print("Dequeued:", queue.pop(0))

while True:
    print("\n1. ENQUEUE")
    print("2. DEQUEUE")
    print("3. DISPLAY")
    print("4. EXIT")

    choice = int(input("Enter choice: "))

    if choice == 1:
        enqueue()
    elif choice == 2:
        dequeue()
    elif choice == 3:
        print("Queue:", queue)
    elif choice == 4:
        break
    else:
        print("Invalid choice")
```

3. Singly Linked List

Write a Python program to implement a Singly Linked List with operations such as create, insert, delete, and traverse.




```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def create(self, data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
        else:
            temp = self.head
            while temp.next:
                temp = temp.next
            temp.next = new_node

    def insert(self, data, position):
        new_node = Node(data)
        if position == 0:
            new_node.next = self.head
            self.head = new_node
            return

        temp = self.head
        for _ in range(position - 1):
            if temp is None:
                return
            temp = temp.next

        if temp is None:
            return

        new_node.next = temp.next
        temp.next = new_node

    def delete(self, key):
        temp = self.head

        if temp and temp.data == key:
            self.head = temp.next
            return

        prev = None
        while temp and temp.data != key:
            prev = temp
            temp = temp.next

        if temp is None:
            return

        prev.next = temp.next

    def traverse(self):
        temp = self.head
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")


sll = SinglyLinkedList()

sll.create(10)
sll.create(20)
sll.create(30)

sll.insert(15, 1)
sll.delete(20)

sll.traverse()
```

4. Real-Life Application

Design and implement a mini project such as a Student Management System or Employee Management System using Python to perform:

Record creation

Data update

Calculations

Report generation

```python

students = {}

def create_record():
    roll = int(input("Enter Roll Number: "))
    name = input("Enter Name: ")
    marks = float(input("Enter Marks: "))

    students[roll] = {"name": name, "marks": marks}
    print("Record Created Successfully")


def update_record():
    roll = int(input("Enter Roll Number to Update: "))

    if roll in students:
        name = input("Enter New Name: ")
        marks = float(input("Enter New Marks: "))

        students[roll]["name"] = name
        students[roll]["marks"] = marks

        print("Record Updated Successfully")
    else:
        print("Record Not Found")


def calculate_average():
    if len(students) == 0:
        print("No records available")
        return

    total = sum(s["marks"] for s in students.values())
    print("Average Marks:", total / len(students))


def generate_report():
    if len(students) == 0:
        print("No records available")
        return

    print("\n----- Student Report -----")
    for roll, data in students.items():
        print("Roll:", roll, "| Name:", data["name"], "| Marks:", data["marks"])


while True:
    print("\n1. Create Record")
    print("2. Update Record")
    print("3. Calculate Average")
    print("4. Generate Report")
    print("5. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        create_record()
    elif choice == 2:
        update_record()
    elif choice == 3:
        calculate_average()
    elif choice == 4:
        generate_report()
    elif choice == 5:
        break
    else:
        print("Invalid Choice")
```

5. Data Structure Selection

Analyze and justify the selection of appropriate linear data structures for different operations used in the mini project.

```python
from collections import deque

students = {}
update_stack = []
processing_queue = deque()


def create_record():
    roll = int(input("Enter Roll No: "))
    name = input("Enter Name: ")
    marks = int(input("Enter Marks: "))

    students[roll] = {"name": name, "marks": marks}
    processing_queue.append(roll)


def update_record():
    roll = int(input("Enter Roll No to update: "))

    if roll in students:
        update_stack.append((roll, students[roll].copy()))

        name = input("Enter new name: ")
        marks = int(input("Enter new marks: "))

        students[roll] = {"name": name, "marks": marks}
        print("Record Updated")
    else:
        print("Record not found")


def generate_report():
    print("\nStudent Report:")
    for roll in students:
        print(roll, students[roll])


def process_queue():
    if processing_queue:
        roll = processing_queue.popleft()
        print("Processed student:", roll)
    else:
        print("Queue is empty")


while True:
    print("\n1.Create 2.Update 3.Report 4.Process Queue 5.Exit")
    ch = int(input("Enter choice: "))

    if ch == 1:
        create_record()
    elif ch == 2:
        update_record()
    elif ch == 3:
        generate_report()
    elif ch == 4:
        process_queue()
    elif ch == 5:
        break
```
